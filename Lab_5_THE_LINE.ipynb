## ⓪ PROJECT SETTINGS — إعدادات المشروع

```python
ASSISTANT_NAME = "THE LINE"

GUESS_WORDS = ["أعتقد", "تقريباً"]

TOLERANCE = 10

print("Assistant :", ASSISTANT_NAME)
print("Guess words:", GUESS_WORDS)
print("Tolerance :", TOLERANCE)
```

## ① GOLDEN SET — المجموعة الذهبية

```python
GOLDEN_SET = [
    {"question": "كيف نبدأ مراجعة كفاءة الطاقة في المشروع؟", "expected": "بيانات الاستهلاك"},
    {"question": "ما الهدف التجريبي لخفض استهلاك الطاقة؟", "expected": "20%"},
    {"question": "كيف نراجع كفاءة استخدام المياه؟", "expected": "بيانات المياه"},
    {"question": "وش أفضل مطعم في تبوك؟", "expected": "خارج نطاق المشروع"}
]

print("Golden cases:", len(GOLDEN_SET))

for case in GOLDEN_SET:
    print(case["question"], "→", case["expected"])
```

## ② answer_question — نسختين من الإجابات

```python
ANSWERS_V1 = {
    "كفاءة الطاقة": "نبدأ مراجعة كفاءة الطاقة من بيانات الاستهلاك الحالية.",
    "الهدف التجريبي": "الهدف التجريبي في هذا التقييم هو خفض الاستهلاك بنسبة 20%.",
    "استخدام المياه": "نبدأ مراجعة كفاءة استخدام المياه من بيانات المياه الحالية."
}

ANSWERS_V2 = {
    "كفاءة الطاقة": "نبدأ مراجعة كفاءة الطاقة من بيانات الاستهلاك الحالية.",
    "الهدف التجريبي": "أعتقد أن الهدف التجريبي تقريباً 30%.",
    "استخدام المياه": "نبدأ مراجعة كفاءة استخدام المياه من بيانات المياه الحالية."
}

def answer_question(question, answers):
    for keyword, answer in answers.items():
        if keyword in question:
            return answer
    return "هذا خارج نطاق المشروع."

print(answer_question("ما الهدف التجريبي لخفض استهلاك الطاقة؟", ANSWERS_V1))
print(answer_question("ما الهدف التجريبي لخفض استهلاك الطاقة؟", ANSWERS_V2))
```

## ③ run_eval — تشغيل التقييم

```python
def run_eval(answers):
    passed = 0

    for case in GOLDEN_SET:
        answer = answer_question(case["question"], answers)

        if case["expected"] in answer:
            passed += 1
            print("PASS |", case["question"])
        else:
            print("FAIL |", case["question"])

    score = passed * 100 // len(GOLDEN_SET)
    print("Score:", score, "%")
    return score

baseline = run_eval(ANSWERS_V1)
```

## ④ demo_judge — النموذج كمُحكِّم

```python
def demo_judge(case, answer):
    if case["expected"] not in answer:
        return 1

    for word in GUESS_WORDS:
        if word in answer:
            return 2

    return 3

case = GOLDEN_SET[1]

print(demo_judge(case, "الهدف التجريبي في هذا التقييم هو خفض الاستهلاك بنسبة 20%."))
print(demo_judge(case, "أعتقد أن الهدف التجريبي هو 20%."))
print(demo_judge(case, "أعتقد أن الهدف التجريبي تقريباً 30%."))
```

## ⑤ REGRESSION GATE — بوابة الانحدار

```python
def gate(new_score, baseline):
    if new_score < baseline - TOLERANCE:
        return "BLOCKED"
    return "PASSED"

new_score = run_eval(ANSWERS_V2)

print("Baseline :", baseline, "%")
print("New score:", new_score, "%")
print("Gate     :", gate(new_score, baseline))
```
